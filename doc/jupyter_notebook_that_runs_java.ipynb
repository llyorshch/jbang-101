## JavaFX Hello World with a parameter

The kernel resolves `//DEPS` lines with JBang and adds them to the classpath. `${os.detected.jfxname}` picks the right native JavaFX build for your OS and architecture.

```java
//DEPS org.openjfx:javafx-controls:25.0.1:${os.detected.jfxname}

import javafx.application.Platform;
import javafx.scene.Scene;
import javafx.scene.control.Label;
import javafx.scene.layout.StackPane;
import javafx.stage.Stage;

void helloFx(String name) {
    try {
        Platform.startup(() -> {});
        Platform.setImplicitExit(false);
    } catch (IllegalStateException alreadyStarted) {
        // JavaFX can only be started once per kernel: reuse it on re-runs
    }
    Platform.runLater(() -> {
        var label = new Label("Hello, " + name + "!");
        label.setStyle("-fx-font-size: 32px;");
        var stage = new Stage();
        stage.setTitle("JavaFX from Jupyter");
        stage.setScene(new Scene(new StackPane(label), 400, 200));
        stage.show();
    });
}
```

Change the parameter and re-run the cell to open a new window:

```java
helloFx("JBang");
```
